```python
from nltk.corpus import brown
import numpy as np
from collections import defaultdict
import tqdm # tqdm displays a progress bar
from tqdm.notebook import tqdm # tqdm is a nice process indicator 

category_vectors = []

cats = brown.categories()
    
# for each category
for cat in cats:
    print(cat)
    # grab all of the documents
    for fileid in tqdm(brown.fileids(categories=[cat])):
        sents = brown.sents(fileids=[fileid])
        sent_vecs = []
        for sent in sents:
            sent = ' '.join(sent)
            sent = nlp(sent)
            # grab all of the words, find their embedding, sum all embeddings
            word_sum = np.sum([tok.vector for tok in sent], axis=0) # why axis=0?
            # add the now summed embedding to the list for this category
            sent_vecs.append(word_sum)
        category_vectors.append((cat,np.sum(sent_vecs, axis=0)))
    
```

##### 3. As you made changes to the network, what do you notice about how hyperparameters (network depth, number of nodes, learning rate, etc.) and how they interact with each other? We said that neural networks are learning non-convex problems, but what about finding the best parameters? Is that a convex problem?

Hyperparameters interact and finding the best combination is a non-convex problem. This means that the performance of the neural network is affected by the combination of hyperparameters (such as learning rate, number of hidden layers, number of neurons in each layer, regularization strength, etc.) in a complex and non-linear way. Changing one hyperparameter can have a ripple effect on the performance of the network, and finding the optimal combination of hyperparameters is a challenging problem.

In other words, the relationship between hyperparameters and network performance is not a simple one-to-one correlation, but rather a complex web of interactions. This makes it difficult to predict how changing one hyperparameter will affect the performance of the network, and requires a careful and iterative approach to find the best combination.

Furthermore, the search space of hyperparameters is often non-convex, meaning that the optimization algorithm may get stuck in a local minimum, rather than finding the global optimum. This requires careful tuning and regularization techniques to avoid overfitting and ensure generalization.

So, hyperparameters like learning rate, number of epochs, and activation functions interact in complex ways. Finding the optimal combination is a non-convex problem, requiring experimentation and iteration.

##### 4. What is regularization? Why is it important?

Regularization is said to prevent the overfitting and is crucial for generalization (also reducing the risk of the model performing poorly on unseen data).

Regularization techniques like Dropout, L1/L2 regularization, and early stopping prevents the network from relying too heavily on any one feature or neuron, ensuring it generalizes well to a new data.

Reducing model complexity: Dropout temporarily drops neurons during training, forcing the network to learn robust features that are not dependent on specific neurons.

Penalizing large weights: L1/L2 penalties add a penalty term to the loss function based on the magnitude of weights, discouraging the model from relying on a few dominant features
.
Early stopping: Stopping training before the model fully memorizes the training data helps prevent overfitting to specific noise patterns.

##### 5. Which activation functions did you choose (besides logitistic/sigmoid)? For one of the activation functions you tried, spend some time learning about it. Whereas logistic/sigmoid maps from inputs to a probability between 0-1, what does the activation function you chose do?

I chose LeakyReLU and softmax activations. LeakyReLU introduces a small negative slope to avoid any dying neurons, while softmax outputs probabilities for multi-class classification. (By using this I got higher accuracy than the other ones)

LeakyReLU: Allows a small amount of the input to pass through, even if it's negative, helping neurons to stay active. Softmax outputs probabilities for each class, allowing for multi-class classification.

Softmax: It gives the output probabilities for each class, allowing for multi-class classification, by transforming the output into a probability distribution where all classes sum to 1.

While various activation functions exist, a common choice for the hidden layers is the ReLU (Rectified Linear Unit) activation.

ReLU: Unlike sigmoid that outputs values between 0 and 1, ReLU outputs the input directly if it's positive and zero otherwise. This introduces non-linearity into the network, allowing it to learn more complex relationships between features.
By applying ReLU, the network can learn to activate only the relevant neurons that contribute to the final prediction, improving the model's ability to differentiate between different topics.

```python
import gensim

from nltk.corpus import brown
import numpy as np
from collections import defaultdict
import tqdm # tqdm displays a progress bar
from tqdm.notebook import tqdm as tqdm # tqdm is a nice process indicator 

category_vectors = []

categ = brown.categories()
    
# for each category
for cat in categ:
    print(cat)
    # grab all of the documents
    for fileid in tqdm(brown.fileids(categories=[cat])):
        sents = brown.sents(fileids=[fileid])
        sent_vecs = []
        for sent in sents:
            # Create CBOW model
            model = gensim.models.Word2Vec([sent], min_count=1, vector_size=100, window=5)
            # grab all of the words, find their embedding, sum all embeddings
            word_sum = np.sum([model.wv[tok] for tok in sent], axis=0) # why axis=0?
            # add the now summed embedding to the list for this category
            sent_vecs.append(word_sum)
        category_vectors.append((cat,np.sum(sent_vecs, axis=0)))
```
